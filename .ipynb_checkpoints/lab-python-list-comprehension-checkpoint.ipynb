````python
def get_customer_orders():
    
    customer_orders = set()
    add_more = "yes"
    
    while add_more == "yes":
        order = input(f"Name 1 product from the list {products}: ")
    
        if order in products:
            customer_orders.add(order)
        else:
            print(f"The item '{order}' is not in the list of products.")
        
        add_more = input("Do you want to add another product (yes/no)? ")

        while add_more not in ["yes", "no"]:
            print("Invalid response")
            add_more = input("Do you want to add another product (yes/no)? ")

    return customer_orders
````
